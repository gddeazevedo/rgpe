```python
import numpy as np


def riemann_advanced_data_map(x):
    x = np.asarray(x)

    x_safe = x + 1e-8

    theta_like = np.sum(x_safe * np.log(x_safe))
    val = np.cos(theta_like)

    for i in range(len(x) - 1):
        val += 0.3 * np.cos(x[i] - x[i+1])
        val += 0.3 * np.sin(x[i] * x[i+1])

    return val


circuit = pauli_feature_map(2, reps=2, paulis=["Y", "YY"], entanglement="full", data_map_func=riemann_advanced_data_map)
print(circuit)
```
